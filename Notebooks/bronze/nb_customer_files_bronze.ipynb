```python
tablename = f"saleslake_{env}.bronze_{env}.rawCustomer"
print(tablename)
srcfile = f"/Volumes/saleslake_{env}/bronze_{env}/vol_saleslake_src_files_{env}/daily_customer/"
print(srcfile)
```

```python
spark.sql (f"""COPY INTO {tablename}
FROM (
    SELECT customer_id,customer_name,email,phone,address,city,state,
    country,zip_code,segment,
    current_timestamp() AS ingest_ts
    From '{srcfile}'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'false')
""")
```
